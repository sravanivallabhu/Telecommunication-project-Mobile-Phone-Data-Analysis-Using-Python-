```python

plt.hist(data['Scr_d'],bins=20)
plt.title('Distribution of Screen Diagonal')
plt.xlabel('Screen Diagonal')
plt.ylabel('Frequency')
plt.show
Q1 = data['Scr_d'].quantile(0.25)
Q2 = data['Scr_d'].quantile(0.50)
Q3 = data['Scr_d'].quantile(0.75)

print("First Quartile (Q1):", Q1)
print("Second Quartile (Q2):", Q2)
print("Third Quartile (Q3):", Q3)
data['Scr_d'].quantile([0.25, 0.50, 0.75])
data[['PC', 'FC']].head()
```

```python
mean_pc = data['PC'].mean()
median_pc = data['PC'].median()

print("Mean:", mean_pc)
print("Median:", median_pc)
sns.histplot(data['PC'], bins=20, kde=True)

plt.axvline(mean_pc, color='red', label='Mean')
plt.axvline(median_pc, color='green', label='Median')

plt.xlabel('Primary Camera (MP)')
plt.ylabel('Frequency')
plt.title('Distribution of Primary Camera')
plt.legend()
plt.show()
```

```python
mean_pc = data['PC'].mean()
mean_fc = data['FC'].mean()
con4 = (data['PC'] >= mean_pc) & (data['FC'] >= mean_fc)
print(con4)
data[con4]
```
